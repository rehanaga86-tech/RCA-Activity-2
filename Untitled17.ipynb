```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.dummy import DummyClassifier
from sklearn.metrics import (
    confusion_matrix, ConfusionMatrixDisplay, accuracy_score,
    precision_score, recall_score, f1_score, roc_auc_score,
    roc_curve, classification_report
)

df = pd.read_csv("/student_exam_performance.csv").drop_duplicates().copy()
print("Shape:", df.shape)
display(df.head())
print("Missing values:\n", df.isnull().sum()[df.isnull().sum() > 0])
```

```python
y = (df["pass_status"] == "Pass").astype(int)
print(y.value_counts())

plt.figure(figsize=(5, 3.5))
sns.countplot(x=y)
plt.xticks([0, 1], ["Fail (0)", "Pass (1)"])
plt.title("Class Distribution (Pass vs Fail)")
plt.xlabel("Class")
plt.ylabel("Count")
plt.show()
```

```python
drop_cols = [
    "student_id", "pass_status", "exam_score",
    "performance_grade", "performance_level",
    "questions_attempted", "questions_correct"
]

X = df.drop(columns=drop_cols)
num_cols = X.select_dtypes(include=["int64", "float64"]).columns.tolist()
cat_cols = X.select_dtypes(include=["object", "category"]).columns.tolist()

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

print("Train size:", len(X_train), "| Test size:", len(X_test))

num_pipe = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

cat_pipe = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("ohe", OneHotEncoder(drop="first", handle_unknown="ignore"))
])

prep = ColumnTransformer([
    ("num", num_pipe, num_cols),
    ("cat", cat_pipe, cat_cols)
])
```

```python
z = np.linspace(-8, 8, 200)
sigmoid = 1 / (1 + np.exp(-z))

plt.figure(figsize=(5.5, 3.5))
plt.plot(z, sigmoid)
plt.axhline(0.5, color="red", linestyle="--", label="Threshold = 0.5")
plt.title("Sigmoid Curve")
plt.xlabel("z")
plt.ylabel("Probability")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
model = Pipeline([
    ("prep", prep),
    ("clf", LogisticRegression(max_iter=1000, random_state=42))
])
model.fit(X_train, y_train)

baseline = DummyClassifier(strategy="most_frequent", random_state=42)
baseline.fit(X_train, y_train)
print("Majority Baseline Accuracy:", round(accuracy_score(y_test, baseline.predict(X_test)), 4))

ohe_cols = model.named_steps["prep"].named_transformers_["cat"].named_steps["ohe"].get_feature_names_out(cat_cols).tolist()
all_cols = num_cols + ohe_cols
coefs = pd.DataFrame({
    "Feature": all_cols,
    "Coef": model.named_steps["clf"].coef_[0]
}).sort_values("Coef", key=abs, ascending=False)

print("Intercept:", round(model.named_steps["clf"].intercept_[0], 4))
display(coefs.head(10).round(4))

probs = model.predict_proba(X_test)[:, 1]
preds = (probs >= 0.5).astype(int)

display(pd.DataFrame({
    "Actual": y_test.values[:10],
    "Prob_Pass": np.round(probs[:10], 4),
    "Pred_0.5": preds[:10]
}))
```

```python
cm = confusion_matrix(y_test, preds)
tn, fp, fn, tp = cm.ravel()

acc = accuracy_score(y_test, preds)
prec = precision_score(y_test, preds)
rec = recall_score(y_test, preds)
spec = tn / (tn + fp)
f1 = f1_score(y_test, preds)
auc = roc_auc_score(y_test, probs)

print(f"TN={tn}, FP={fp}, FN={fn}, TP={tp}")
print(f"Accuracy: {acc:.4f}")
print(f"Precision: {prec:.4f}")
print(f"Recall: {rec:.4f}")
print(f"Specificity: {spec:.4f}")
print(f"F1-Score: {f1:.4f}")
print(f"ROC-AUC: {auc:.4f}")
print("Train Accuracy:", round(accuracy_score(y_train, model.predict(X_train)), 4))
print("\n", classification_report(y_test, preds, target_names=["Fail", "Pass"], digits=4))

fig, ax = plt.subplots(1, 2, figsize=(10, 4))
ConfusionMatrixDisplay(cm, display_labels=["Fail", "Pass"]).plot(ax=ax[0], cmap="Blues", colorbar=False)
ax[0].set_title("Confusion Matrix")

fpr, tpr, _ = roc_curve(y_test, probs)
ax[1].plot(fpr, tpr, label=f"AUC = {auc:.4f}")
ax[1].plot([0, 1], [0, 1], linestyle="--", color="gray")
ax[1].set_title("ROC Curve")
ax[1].set_xlabel("False Positive Rate")
ax[1].set_ylabel("True Positive Rate")
ax[1].legend()
plt.tight_layout()
plt.show()
```

```python
table = []
for t in [0.3, 0.5, 0.7]:
    p = (probs >= t).astype(int)
    table.append({
        "Threshold": t,
        "Pred_Positive": int(p.sum()),
        "Accuracy": round(accuracy_score(y_test, p), 4),
        "Precision": round(precision_score(y_test, p), 4),
        "Recall": round(recall_score(y_test, p), 4),
        "F1": round(f1_score(y_test, p), 4)
    })

display(pd.DataFrame(table))
```

```python
feat = "previous_exam_score"

plt.figure(figsize=(6, 4))
plt.scatter(X_test[feat], probs, c=y_test, cmap="coolwarm", alpha=0.3, s=12)
plt.axhline(0.5, color="black", linestyle="--", label="Threshold = 0.5")
plt.title(f"Predicted Pass Probability vs {feat}")
plt.xlabel(feat)
plt.ylabel("P(Pass = 1)")
plt.legend()
plt.show()

new_row = {c: float(X_train[c].median()) for c in num_cols}
for c in cat_cols:
    new_row[c] = X_train[c].mode()[0]

new_row["previous_exam_score"] = 76.0
new_row["attendance_percentage"] = 88.0

new_df = pd.DataFrame([new_row])
new_prob = model.predict_proba(new_df)[0, 1]
new_pred = int(new_prob >= 0.5)

print("New student probability:", round(new_prob, 4))
print("New student prediction:", "Pass" if new_pred == 1 else "Fail")
```
